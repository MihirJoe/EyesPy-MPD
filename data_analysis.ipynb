```python
import pandas as pd
```

## Add MRD1 (pupil to upper)

```python
df = pd.read_csv("eye_frames_20250414_124415/eye_measurements.csv")
df.head()
```

```python
df['right_pupil_to_upper'] = df['right_palpebral_height'] - df['right_pupil_to_lower']
df['left_pupil_to_upper'] = df['left_palpebral_height'] - df['left_pupil_to_lower']
```

```python
df.head()
```

```python
df.drop(columns=['right_pupil_to_lower', 'left_pupil_to_lower'], inplace=True)
```

```python
df.head()
```

```python
df['timestamp'] -= df['timestamp'].iloc[0]
df.head()
```

```python
len(df)
```

```python
df.to_csv("eye_frames_20250414_124415/actual_eye_measurements.csv", index=False)
```
